## **Linear Regresison**

## **Load Required Libraries**

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import statsmodels.api as sm 
```

## **Load Data**

```python
salary_df = pd.read_csv("data/Salary_dataset.csv")
```

```python
salary_df.head()
```

```python
salary_df.tail()
```

```python
salary_df.columns
```

## **Instantiate and fit the model**
 - The dependent variable is a continuous numeric variable

```python
X = sm.add_constant(salary_df["YearsExperience"])
y = salary_df["Salary"]

model = sm.OLS(y, X).fit()
print(model.summary())
```

## ** evaluation **

```python
# Predict probablities for a range of values
# Predict fitted salary values for a range of experience levels
experience_range = np.linspace(0, 10, 100)
X_pred = sm.add_constant(experience_range)

pred_salary = model.predict(X_pred)
```

 - experience_range = Years of Experience
 - pred_salary = predicted Salary (continuous)

```python
# scatter plot : 

plt.figure(figsize=(10, 6))

# Observed data
plt.scatter(
    salary_df["YearsExperience"],
    salary_df["Salary"],
    alpha=0.6,
    label="Observed"
)

# Fitted regression line
plt.plot(
    experience_range,
    pred_salary,
    color="red",
    linewidth=2,
    label="Fitted OLS line"
)

plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("OLS Regression: Salary vs Years of Experience")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

```

## ** Descriptive Interpretation**
 - X‑axis: Years of Experience
 - Y‑axis: Salary
 - Blue points (Observed): Actual observed salary values at different experience levels
 - Red line (Fitted Ordinary Least Squares (OLS) line): The estimated linear relationship between Years of Experience and Salary obtained from the OLS regression

 - Visually, the red line passes through the center of the data cloud, indicating a strong linear association.

 - The figure shows a strong positive linear relationship between Years of Experience and Salary:
 - As years of experience increase, salary increases in a steady and approximately linear manner
 - Individuals with fewer years of experience earn lower salaries
 - Individuals with more years of experience earn substantially higher salaries
 - This pattern is consistent across the full observed range of experience (roughly 1–11 years).
